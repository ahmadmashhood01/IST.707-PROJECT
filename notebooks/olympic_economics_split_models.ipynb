## 2. WDI structure, coverage, and cleaning

Coverage is calculated across **all rows in the merged file**, before choosing any analysis window. Drop only WDI indicators with less than 10% nonmissing data; retain exactly 10%. Identifiers, Olympic fields, zero values, and all rows—including historical countries—are preserved.

```python
ID_COLS = ["ID", "Country", "wdi_code", "year", "country", "olympic_code",
           "athlete_count", "event_entries", "gold", "silver", "bronze", "total"]
required = {"wdi_code", "year", "country", "olympic_code", "athlete_count",
            "event_entries", "gold", "silver", "bronze", "total"}
missing = required.difference(data.columns)
if missing:
    raise ValueError(f"Missing required merged-dataset columns: {sorted(missing)}")
WDI_COLS = [c for c in data.columns if c not in ID_COLS]
data[WDI_COLS] = data[WDI_COLS].replace(r"^\s*$|^\s*\.\.\s*$", np.nan, regex=True)
coverage = data[WDI_COLS].notna().mean()
DROP_COLS = coverage[coverage < 0.10].index.tolist()
ECON_COLS = coverage[coverage >= 0.10].index.tolist()
coverage_report = pd.DataFrame({"pct_filled": coverage * 100,
                                "pct_missing": (1 - coverage) * 100,
                                "action": np.where(coverage < 0.10, "drop", "keep")})
data = data.drop(columns=DROP_COLS)
assert len(data) == original_shape[0]
print("WDI indicators before / dropped / retained:", len(WDI_COLS), len(DROP_COLS), len(ECON_COLS))
print("Cleaned shape:", data.shape)
print("Years:", data["year"].min(), "to", data["year"].max())
print("WDI codes:", data["wdi_code"].nunique())
display(coverage_report.sort_values("pct_filled"))
print("Dropped WDI indicators:")
for col in DROP_COLS:
    print(f"{col}: {coverage[col] * 100:.2f}% filled")
miss = (data[ECON_COLS].isna().mean() * 100).sort_values(ascending=False)

olympics = data[data["olympic_code"].notna()].copy()
panel = olympics[olympics["year"].between(1990, 2016)].copy()
```

## 9. Stakeholder metrics (2016 snapshot)


```python
snap = m[m["year"] == 2016].copy()
snap = snap.dropna(subset=["log_GDP (current US$)", "total"])
snap["GDP (current US$)"] = 10 ** snap["log_GDP (current US$)"]
snap["log_gdp"] = np.log10(snap["GDP (current US$)"])
snap["medals_per_log_gdp"] = snap["total"] / snap["log_gdp"]

display(
    snap[snap["total"] > 0]
    .nlargest(10, "medals_per_log_gdp")[
        ["country", "athlete_count", "total", "GDP (current US$)", "medals_per_log_gdp"]
    ]
)

if len(snap) < 2 or snap["log_gdp"].nunique() < 2:
    print("Insufficient complete observations for the GDP baseline.")
else:
    coefs = np.polyfit(snap["log_gdp"], snap["total"], 1)
    snap["expected"] = np.polyval(coefs, snap["log_gdp"])
    snap["residual"] = snap["total"] - snap["expected"]
    display(snap.nlargest(8, "residual")[["country", "total", "expected", "residual"]])
    display(snap.nsmallest(8, "residual")[["country", "total", "expected", "residual"]])
```

### 12. Using different metrics to compare Olympic success

```python
country_stats = (
    olympics.groupby(["country", "olympic_code"])[["total", "athlete_count"]]
    .sum(min_count=1)
    .rename(columns={
        "total": "medals",
        "athlete_count": "athlete_participations"
    })
)

matched = olympics.dropna(subset=["total", "athlete_count"])

country_stats = (
    matched.groupby(["country", "olympic_code"])[["total", "athlete_count"]]
    .sum()
    .rename(columns={
        "total": "medals",
        "athlete_count": "athlete_participations"
    })
)

country_stats = country_stats[
    country_stats["athlete_participations"] > 0
].copy()

country_stats["medals_per_athlete"] = (
    country_stats["medals"] / country_stats["athlete_participations"]
)

top15 = (
    country_stats
    .sort_values(
        ["medals_per_athlete", "medals"],
        ascending=[False, False]
    )
    .head(15)
    .reset_index()
)

display(top15)

fig, axes = plt.subplots(1, 3, figsize=(18, 8), sharey=True)

plots = [
    ("medals", "Total medals", "goldenrod"),
    ("athlete_participations", "Athlete participations", "steelblue"),
    ("medals_per_athlete", "Medals per athlete participation", "seagreen"),
]

for ax, (column, title, color) in zip(axes, plots):
    ax.barh(top15["country"], top15[column], color=color)
    ax.set_title(title)
    ax.set_xlabel(title)
    ax.grid(axis="y", visible=False)

axes[0].invert_yaxis()

fig.suptitle(
    "Top 15 countries by medals per athlete participation "
    f"({matched['year'].min():.0f}–{matched['year'].max():.0f})",
    fontsize=15
)
plt.tight_layout()
plt.show()
```

### 13. Finding the best year to track current GDP per Capita and Population

```python
TARGET_YEAR = 2016

variables = {
    "Population, total": "population_year",
    "GDP per capita (current US$)": "gdp_per_capita_year"
}

eligible = data[
    data["wdi_code"].notna() & (data["year"] <= TARGET_YEAR)
].copy()

country_2016 = (
    eligible.sort_values("year")
    .groupby("wdi_code")[["Country"]]
    .last()
)

for column, year_column in variables.items():
    values = eligible[["wdi_code", "year", column]].copy()
    values[column] = pd.to_numeric(values[column], errors="coerce")

    latest = (
        values
        .dropna(subset=[column])
        .loc[lambda df: df[column] > 0]
        .sort_values("year")
        .drop_duplicates("wdi_code", keep="last")
        .set_index("wdi_code")
        .rename(columns={"year": year_column})
    )

    country_2016 = country_2016.join(
        latest[[column, year_column]]
    )

for column, year_column in variables.items():
    print(f"\n{column}")
    print("Using 2016:", country_2016[year_column].eq(2016).sum())
    print("Using earlier years:", country_2016[year_column].lt(2016).sum())
    print("No available data:", country_2016[column].isna().sum())

    display(
        country_2016[year_column]
        .value_counts()
        .sort_index(ascending=False)
        .rename_axis("source_year")
        .to_frame("countries")
    )
```

### 14. Using trial and error to split countries into different groups

```python
grouped_countries = country_2016.copy()

population_labels = [
    "Under 1M", "1–under 15M", "15–under 50M",
    "50M+"
]

gdp_labels = [
    "Under $2.5K", "$2.5K–under $15K", "$15K–under $30K",
    "$30K+"
]

grouped_countries["population_group"] = pd.cut(
    grouped_countries["Population, total"],
    bins=[0, 1e6, 15e6, 50e6, np.inf],
    labels=population_labels,
    right=False
)

grouped_countries["gdp_per_capita_group"] = pd.cut(
    grouped_countries["GDP per capita (current US$)"],
    bins=[0, 2500, 15000, 40000, np.inf],
    labels=gdp_labels,
    right=False
)

valid = grouped_countries.dropna(
    subset=["population_group", "gdp_per_capita_group"]
)

counts = pd.crosstab(
    valid["population_group"],
    valid["gdp_per_capita_group"],
    dropna=False
).reindex(
    index=population_labels,
    columns=gdp_labels,
    fill_value=0
)

if valid.empty:
    raise ValueError("No countries have both population and GDP per capita.")

percentages = counts / len(valid) * 100

annotations = counts.copy().astype(str)

for row in counts.index:
    for col in counts.columns:
        annotations.loc[row, col] = (
            f"{counts.loc[row, col]}\n"
            f"({percentages.loc[row, col]:.1f}%)"
        )

plt.figure(figsize=(12, 7))
sns.heatmap(
    counts,
    annot=annotations,
    fmt="",
    cmap="Blues",
    linewidths=0.5,
    cbar_kws={"label": "Number of countries"}
)

plt.xlabel("GDP per capita group")
plt.ylabel("Population group")
plt.title(
    "Country groups by population and GDP per capita\n"
    "2016 values with earlier-year fallback"
)
plt.xticks(rotation=25, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

group_members = (
    valid.reset_index()
    .assign(
        country_name=lambda df: df["Country"].fillna(df["wdi_code"])
    )
    .groupby(
        ["population_group", "gdp_per_capita_group"],
        observed=True
    )
    .agg(
        country_count=("wdi_code", "size"),
        countries=("country_name", lambda names: ", ".join(sorted(names)))
    )
    .reset_index()
)
```

b.Data to help refine groups, to find where there are clearly tiers between groups

```python
group_names = (
    group_members[
        ["population_group", "gdp_per_capita_group"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

group_names["group_name"] = [
    f"Group {i + 1}" for i in range(len(group_names))
]

country_groups = (
    valid.reset_index()
    .merge(
        group_names,
        on=["population_group", "gdp_per_capita_group"],
        how="left",
        validate="many_to_one"
    )
    .rename(columns={
        "Country": "country",
        "Population, total": "population",
        "GDP per capita (current US$)": "gdp_per_capita"
    })
)

country_groups["country"] = (
    country_groups["country"].fillna(country_groups["wdi_code"])
)

country_groups = (
    country_groups[
        [
            "group_name",
            "wdi_code",
            "country",
            "population_group",
            "gdp_per_capita_group",
            "population",
            "gdp_per_capita",
            "population_year",
            "gdp_per_capita_year"
        ]
    ]
    .sort_values(
        ["population_group", "gdp_per_capita_group", "country"]
    )
    .reset_index(drop=True)
)
display(country_groups)
```

### 15. Finding missing economic facotrs for each group

```python
economic_factors = ECON_COLS

economics_2016 = data.loc[
    data["year"].eq(2016) & data["wdi_code"].notna(),
    ["wdi_code"] + economic_factors
].copy()

group_economics = country_groups[
    ["wdi_code", "country", "group_name"]
].merge(
    economics_2016,
    on="wdi_code",
    how="left",
    validate="one_to_one"
)

group_order = group_names["group_name"].tolist()

included_by_group = (
    group_economics[economic_factors]
    .notna()
    .groupby(group_economics["group_name"])
    .mean()
    .mul(100)
    .reindex(group_order)
)

included_by_group.index.name = "group_name"

group_summary = pd.DataFrame({
    "countries": (
        group_economics.groupby("group_name")
        .size()
        .reindex(group_order)
    ),
    "overall_included_pct": included_by_group.mean(axis=1)
})

levels = ["Very low", "Low", "Middle", "High"]

group_numbers = (
    group_economics["group_name"]
    .str.extract(r"(\d+)$")[0]
    .astype(int)
)

if not group_numbers.between(1, 16).all():
    raise ValueError("This mapping requires group numbers from 1 to 16.")

chart_data = group_economics.copy()

chart_data["population_category"] = (
    ((group_numbers - 1) // 4).map(dict(enumerate(levels)))
)

chart_data["gdp_category"] = (
    ((group_numbers - 1) % 4).map(dict(enumerate(levels)))
)

chart_data["included_pct"] = (
    chart_data[economic_factors].notna().mean(axis=1) * 100
)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
summaries = {}

for ax, category, title in [
    (axes[0], "population_category", "Population"),
    (axes[1], "gdp_category", "GDP per capita")
]:
    summary = (
        chart_data.groupby(category)
        .agg(
            countries=("wdi_code", "nunique"),
            included_pct=("included_pct", "mean")
        )
        .reindex(levels)
    )

    summary["countries"] = summary["countries"].fillna(0).astype(int)
    summaries[title] = summary

    bars = ax.bar(
        levels,
        summary["included_pct"],
        color="steelblue" if title == "Population" else "seagreen"
    )

    ax.bar_label(
        bars,
        labels=[
            f"{pct:.1f}%\n{count} countries"
            if pd.notna(pct) else "No countries"
            for pct, count in zip(
                summary["included_pct"], summary["countries"]
            )
        ],
        padding=5
    )

    ax.set_title(f"Economic data included by {title.lower()}")
    ax.set_ylabel("Data included (%)")
    ax.set_ylim(0, 115)
    ax.grid(axis="x", visible=False)

plt.tight_layout()
plt.show()

display(
    pd.concat(
        summaries,
        names=["grouped_by", "category"]
    ).round(1)
)
```

### 16. Evidence for dividing into groups

```python
results_2016 = data.loc[
    data["year"].eq(2016) & data["wdi_code"].notna(),
    ["wdi_code", "total", "athlete_count"]
].copy()

medal_data = chart_data[
    [
        "wdi_code", "country", "group_name",
        "population_category", "gdp_category"
    ]
].merge(
    results_2016,
    on="wdi_code",
    how="left",
    validate="one_to_one"
)

medal_data = medal_data.dropna(subset=["total", "athlete_count"])
medal_data = medal_data[medal_data["athlete_count"] > 0].copy()

medal_data["medals_per_participation"] = (
    medal_data["total"] / medal_data["athlete_count"]
)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

for ax, category, title in [
    (axes[0], "population_category", "Population"),
    (axes[1], "gdp_category", "GDP per capita")
]:
    summary = (
        medal_data.groupby(category)
        .agg(
            average_ratio=("medals_per_participation", "mean"),
            countries=("wdi_code", "nunique")
        )
        .reindex(levels)
    )

    bars = ax.bar(
        levels,
        summary["average_ratio"],
        color="steelblue" if title == "Population" else "seagreen"
    )

    ax.bar_label(
        bars,
        labels=[
            f"{ratio:.3f}\n{int(count)} countries"
            if pd.notna(ratio) else "No data"
            for ratio, count in zip(
                summary["average_ratio"], summary["countries"]
            )
        ],
        padding=4
    )

    ax.set_title(f"Average medals per participation by {title.lower()}")
    ax.set_ylabel("Average country medals / athletes")
    ax.margins(y=0.25)
    ax.grid(axis="x", visible=False)

plt.tight_layout()
plt.show()

group_order_16 = [f"Group {i}" for i in range(1, 17)]

group_medals = (
    medal_data.groupby("group_name")
    .agg(
        average_medals_per_participation=(
            "medals_per_participation", "mean"
        ),
        countries=("wdi_code", "nunique"),
        total_medals=("total", "sum"),
        total_participations=("athlete_count", "sum")
    )
    .reindex(group_order_16)
)

group_medals["countries"] = (
    group_medals["countries"].fillna(0).astype(int)
)

fig, ax = plt.subplots(figsize=(15, 6))

bars = ax.bar(
    group_medals.index,
    group_medals["average_medals_per_participation"],
    color="darkorange"
)

ax.bar_label(
    bars,
    labels=[
        f"{ratio:.3f}\n(n={count})"
        if pd.notna(ratio) else "No data"
        for ratio, count in zip(
            group_medals["average_medals_per_participation"],
            group_medals["countries"]
        )
    ],
    padding=4
)

ax.set_title("2016: average medals per participation in each group")
ax.set_ylabel("Average country medals / athletes")
ax.set_xlabel("Group")
ax.tick_params(axis="x", rotation=45)
ax.margins(y=0.25)
ax.grid(axis="x", visible=False)

plt.tight_layout()
plt.show()
```

### 18. Assigning loose countries to groups to have better data for Machine Learning process

### 19. Machine Learing without groups vs. with group to predict medal success with economic factors

### 19A. Prepare 2016 medal-count data and select well-covered indicators

Keep economic indicators with at least 80% observed values among eligible 2016 countries. Both models use the same countries and train/test split.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

model_data = medal_data[["wdi_code", "country", "group_name", "total"]].merge(
    group_economics[["wdi_code"] + ECON_COLS], on="wdi_code", how="left", validate="one_to_one"
)
model_data[ECON_COLS] = model_data[ECON_COLS].apply(pd.to_numeric, errors="coerce")
model_data["total"] = pd.to_numeric(model_data["total"], errors="coerce")
model_data = model_data.dropna(subset=["total", "group_name"]).copy()

MIN_COVERAGE = 0.80
feature_coverage = model_data[ECON_COLS].notna().mean().sort_values(ascending=False)
selected_cols = feature_coverage[feature_coverage >= MIN_COVERAGE].index.tolist()
excluded_cols = [c for c in selected_cols if "gdp" in c.lower() or "population" in c.lower()]
reduced_cols = [c for c in selected_cols if c not in excluded_cols]
if not selected_cols or not reduced_cols:
    raise ValueError("No usable economic predictors meet the coverage threshold; lower MIN_COVERAGE.")

coverage_table = pd.DataFrame({"coverage_pct": feature_coverage.mul(100).round(1),
                               "used": feature_coverage.ge(MIN_COVERAGE)})
display(coverage_table)
print(f"Countries: {len(model_data)} | All factors: {len(selected_cols)} | Without GDP/population: {len(reduced_cols)}")
train_idx, test_idx = train_test_split(np.arange(len(model_data)), test_size=0.25, random_state=42)
y = model_data["total"]
```

### 19B. Economic correlations with total medal count

```python
correlations = model_data[selected_cols].corrwith(y).dropna()
correlations = correlations.sort_values(key=lambda x: x.abs(), ascending=False)
display(correlations.head(20).rename("pearson_r").to_frame())
correlations.head(20).sort_values().plot.barh(figsize=(10, 8), title="Top 20 economic correlations with total medals")
plt.xlabel("Pearson correlation (r)")
plt.tight_layout()
plt.show()
```

### 19C. Model 1 — All well-covered economic factors

```python
def fit_medal_model(columns, use_groups=False):
    numeric = Pipeline([("impute", SimpleImputer(strategy="median", keep_empty_features=True)),
                        ("scale", StandardScaler())])
    transforms = [("numeric", numeric, columns)]
    features = list(columns)
    if use_groups:
        transforms.append(("groups", OneHotEncoder(handle_unknown="ignore"), ["group_name"]))
        features.append("group_name")
    pipeline = Pipeline([("preprocess", ColumnTransformer(transforms)), ("ridge", Ridge(alpha=100.0))])
    pipeline.fit(model_data.iloc[train_idx][features], y.iloc[train_idx])
    predictions = pipeline.predict(model_data.iloc[test_idx][features])
    actual = y.iloc[test_idx]
    metrics = {"r2": r2_score(actual, predictions),
               "mae": mean_absolute_error(actual, predictions),
               "rmse": np.sqrt(mean_squared_error(actual, predictions))}
    return pipeline, predictions, metrics

all_model, pred_all, all_metrics = fit_medal_model(selected_cols)
display(pd.DataFrame([all_metrics], index=["All economic factors"]).round(3))
plt.figure(figsize=(6, 5))
plt.scatter(y.iloc[test_idx], pred_all, alpha=0.7)
lims = [min(y.iloc[test_idx].min(), pred_all.min()), max(y.iloc[test_idx].max(), pred_all.max())]
plt.plot(lims, lims, "k--")
plt.xlabel("Actual total medals")
plt.ylabel("Predicted total medals")
plt.title("All economic factors")
plt.tight_layout()
plt.show()
```

### 19D. Model 2 — Country groups, excluding GDP and population indicators

```python
group_model, pred_group, group_metrics = fit_medal_model(reduced_cols, use_groups=True)
display(pd.DataFrame([group_metrics], index=["Groups, no GDP/population factors"]).round(3))
plt.figure(figsize=(6, 5))
plt.scatter(y.iloc[test_idx], pred_group, alpha=0.7)
lims = [min(y.iloc[test_idx].min(), pred_group.min()), max(y.iloc[test_idx].max(), pred_group.max())]
plt.plot(lims, lims, "k--")
plt.xlabel("Actual total medals")
plt.ylabel("Predicted total medals")
plt.title("Groups without GDP/population factors")
plt.tight_layout()
plt.show()
```

### 19E. Side-by-side model comparison

```python
comparison = pd.DataFrame([all_metrics, group_metrics],
    index=["All economic factors", "Groups, no GDP/population factors"])
display(comparison.round(3))
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
actual = y.iloc[test_idx].to_numpy()
for ax, prediction, name in zip(axes, [pred_all, pred_group], comparison.index):
    ax.scatter(actual, prediction, alpha=0.7)
    bounds = [min(actual.min(), prediction.min()), max(actual.max(), prediction.max())]
    ax.plot(bounds, bounds, "k--", linewidth=1)
    ax.set(xlabel="Actual total medals", ylabel="Predicted total medals", title=name)
plt.tight_layout()
plt.show()
print("Test countries:", len(test_idx), "| Minimum indicator coverage:", f"{MIN_COVERAGE:.0%}")
print("Groups are defined using GDP per capita and population, so Model 2 still contains indirect information from them.")
```

## 20. Midterm deliverables: null baselines, classification, and tree ensembles

We hold out 25% of 2016 countries (same split as §19). Regression targets **total medals**; classification targets **won any medal** (total > 0). Null baselines show the floor a serious model must beat.

### 20A. Null / random baselines (regression and classification)

```python
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, f1_score

y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
y_cls_train = (y_train > 0).astype(int)
y_cls_test = (y_test > 0).astype(int)

mean_pred = np.full(len(y_test), y_train.mean())
rng = np.random.default_rng(42)
rand_pred = rng.uniform(y_train.min(), y_train.max(), size=len(y_test))

def reg_metrics(y_true, pred):
    return {
        "r2": r2_score(y_true, pred),
        "mae": mean_absolute_error(y_true, pred),
        "rmse": np.sqrt(mean_squared_error(y_true, pred)),
    }

null_rows = [
    reg_metrics(y_test, mean_pred),
    reg_metrics(y_test, rand_pred),
]
null_df = pd.DataFrame(null_rows, index=["Null: predict train mean", "Null: random uniform"]).round(3)
display(null_df)

majority = int(y_cls_train.mode().iloc[0])
maj_pred = np.full(len(y_cls_test), majority)
print(
    "Null classifier (always no medal):",
    f"accuracy={accuracy_score(y_cls_test, maj_pred):.3f}, f1={f1_score(y_cls_test, maj_pred, zero_division=0):.3f}",
)
```

### 20B. Logistic regression — any medal (with confusion matrix)

Stakeholders often care whether a country reaches the podium at all; F1 balances precision and recall when many countries win zero medals.

```python
from sklearn.linear_model import LogisticRegression

log_pipe = Pipeline([
    ("preprocess", ColumnTransformer([
        ("numeric", Pipeline([
            ("impute", SimpleImputer(strategy="median", keep_empty_features=True)),
            ("scale", StandardScaler()),
        ]), reduced_cols),
        ("groups", OneHotEncoder(handle_unknown="ignore"), ["group_name"]),
    ])),
    ("model", LogisticRegression(max_iter=2000, class_weight="balanced")),
])
log_features = list(reduced_cols) + ["group_name"]
log_pipe.fit(model_data.iloc[train_idx][log_features], y_cls_train)
cls_pred = log_pipe.predict(model_data.iloc[test_idx][log_features])
print(f"Accuracy: {accuracy_score(y_cls_test, cls_pred):.3f} | F1: {f1_score(y_cls_test, cls_pred, zero_division=0):.3f}")
cm = confusion_matrix(y_cls_test, cls_pred, labels=[0, 1])
ConfusionMatrixDisplay(cm, display_labels=["No medal", "Any medal"]).plot(cmap="Blues")
plt.title("Logistic regression — any medal (2016 test set)")
plt.tight_layout()
plt.show()
```

### 20C. Decision tree, random forest, and XGBoost (total medals)

Tree models capture nonlinear relationships among health, education, and trade indicators without hand-specified interactions.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor

try:
    from xgboost import XGBRegressor
except ImportError:
    XGBRegressor = None

def fit_tree_regressor(estimator, columns, use_groups=False):
    numeric = Pipeline([
        ("impute", SimpleImputer(strategy="median", keep_empty_features=True)),
        ("scale", StandardScaler()),
    ])
    transforms = [("numeric", numeric, columns)]
    features = list(columns)
    if use_groups:
        transforms.append(("groups", OneHotEncoder(handle_unknown="ignore"), ["group_name"]))
        features.append("group_name")
    pipe = Pipeline([("preprocess", ColumnTransformer(transforms)), ("model", estimator)])
    pipe.fit(model_data.iloc[train_idx][features], y.iloc[train_idx])
    pred = pipe.predict(model_data.iloc[test_idx][features])
    return pipe, pred, reg_metrics(y_test, pred)

dt_pipe, pred_dt, dt_metrics = fit_tree_regressor(
    DecisionTreeRegressor(max_depth=6, random_state=42), reduced_cols, use_groups=True
)
rf_pipe, pred_rf, rf_metrics = fit_tree_regressor(
    RandomForestRegressor(n_estimators=200, max_depth=8, random_state=42), reduced_cols, use_groups=True
)
xgb_metrics = None
pred_xgb = None
if XGBRegressor is not None:
    xgb_pipe, pred_xgb, xgb_metrics = fit_tree_regressor(
        XGBRegressor(n_estimators=200, max_depth=4, learning_rate=0.1, random_state=42,
                     objective="reg:squarederror"),
        reduced_cols,
        use_groups=True,
    )

rows = [all_metrics, group_metrics, dt_metrics, rf_metrics]
names = ["Ridge: all factors", "Ridge: groups, no GDP/pop", "Decision tree", "Random forest"]
if xgb_metrics:
    rows.append(xgb_metrics)
    names.append("XGBoost")
rows.extend([null_rows[0]])
names.append("Null: train mean")
midterm_table = pd.DataFrame(rows, index=names).round(3)
display(midterm_table)
```

### 20D. Feature importance (random forest)

```python
pre = rf_pipe.named_steps["preprocess"]
model = rf_pipe.named_steps["model"]
feature_names = list(reduced_cols)
if "groups" in pre.named_transformers_:
    ohe = pre.named_transformers_["groups"]
    feature_names += list(ohe.get_feature_names_out(["group_name"]))
imp = pd.Series(model.feature_importances_, index=feature_names).sort_values(ascending=False)
imp.head(15).plot.barh(figsize=(8, 5), title="Top 15 random forest importances")
plt.xlabel("Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
